<h1>Bank Account Management System</h1>

```python
class BankAccount:
    def __init__ (self, holder, balance):
        self.holder = holder
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print("Deposit Successful !")
        else:
            print("Enter a Valid Amount")

    def withdraw(self, amount):
        if amount <= 0:
            print("Enter a Valid Withdrawal Amount")

        elif amount > self.balance:
            print("Insufficient Balance!")

        else:
            self.balance -= amount
            print("Withdrawal Successful!")

    def fixed_deposit(self, amount, percent, years):
        if amount <= 0 or percent < 0 or years <= 0:
            print("Enter Valid FD Details")
        elif amount > self.balance:
            print("Insufficient Balance for FD!")
        else:
            self.balance -= amount

            interest = amount * percent * years / 100
            maturity = amount + interest

            print("FD Successful!")
            print("FD Amount :",amount)
            print("Interest Rate :", percent, "%")
            print("Duration :",years, "years")
            print("Interest Earned :", interest)
            print("Maturity Amount :", maturity)
        
    def show_deposit(self):
        print("Account Holder :",self.holder)
        print("Balance :",self.balance)

Account1 = BankAccount("Karthik", 250000)
Account1.deposit(100000)
Account1.withdraw(500)
Account1.show_deposit()
Account1.fixed_deposit(5000, 7, 2)
```
